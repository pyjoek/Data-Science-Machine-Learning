Array Manipulation: Create a 3D NumPy array of shape (4, 3, 2) with random integers between 0 and 10. Then, extract the sub-array that contains the second and third layers.

```python
import numpy as np

# Create a 3D array of shape (4, 3, 2) with random integers between 0 and 10
array_3d = np.random.randint(0, 11, size=(4, 3, 2))
print("Original array:\n", array_3d)

# Extract the second and third layers (index 1 and 2)
sub_array = array_3d[1:3]
print("Extracted layers:\n", sub_array)
```

Broadcasting: Given two arrays A of shape (3, 1) and B of shape (1, 4), perform an element-wise multiplication using broadcasting and display the result.

```python
# Arrays
A = np.array([[1], [2], [3]])  # shape (3,1)
B = np.array([[4, 5, 6, 7]])   # shape (1,4)

# Multiplying with broadcasting
result = A * B
print("Result:\n", result)
```

Advanced Indexing: Given a 1D array of 20 random integers between 0 and 100, find the indices of all elements greater than 50 and replace those elements with -1.

```python
# Generate array
arr = np.random.randint(0, 101, size=20)
print("Original array:\n", arr)

# Find indices where elements > 50
indices = np.where(arr > 50)
# Replace those elements
arr[indices] = -1
print("Modified array:\n", arr)
```

DataFrame Creation: Create a DataFrame with 10 rows and 4 columns (A, B, C, D) with random values. Set A as the index.

```python
import pandas as pd
import numpy as np

np.random.seed(0)  # for reproducibility
df = pd.DataFrame(np.random.randn(10, 4), columns=['A', 'B', 'C', 'D'])
df.set_index('A', inplace=True)  # set 'A' as index
print(df)
```

Groupby & Aggregation: Using the DataFrame, group by column B and compute the mean and sum of columns C and D.

```python
# For demonstration, reset index and create a sample 'B' column
df.reset_index(inplace=True)
# Create a sample 'B' column
df['B'] = np.random.choice(['X', 'Y', 'Z'], size=10)

# Group by 'B'
grouped = df.groupby('B')[['C', 'D']].agg(['mean', 'sum'])
print(grouped)
```

Data Cleaning: Given a DataFrame with missing values, fill the missing values with the mean of their respective columns.

```python
import pandas as pd
import numpy as np

# Original DataFrame
np.random.seed(0)
df = pd.DataFrame(np.random.randn(10, 4), columns=['A', 'B', 'C', 'D'])

# Introduce missing values
df_missing = df.copy()
df_missing.loc[0:2, 'C'] = np.nan
df_missing.loc[4:6, 'D'] = np.nan

# Fill missing values in numeric columns
numeric_cols = df_missing.select_dtypes(include=[np.number]).columns
df_filled = df_missing.copy()

for col in numeric_cols:
    mean_value = df_missing[col].mean()
    df_filled[col].fillna(mean_value, inplace=True)

print(df_filled)
```

Line Plot: Plot a sine wave and a cosine wave on the same figure with labels, legends, and grid.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 100)
sin_y = np.sin(x)
cos_y = np.cos(x)

plt.plot(x, sin_y, label='Sine')
plt.plot(x, cos_y, label='Cosine')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Sine and Cosine Waves')
plt.legend()
plt.grid(True)
plt.show()
```

Histogram: Generate 1000 random samples from a normal distribution and plot their histogram with appropriate labels and title.

```python
samples = np.random.randn(1000)

plt.hist(samples, bins=30, alpha=0.7)
plt.title('Histogram of Normal Distribution Samples')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

Subplots: Create a 2x2 grid of subplots showing different mathematical functions: sine, cosine, tangent, and exponential.

```python
x = np.linspace(-2*np.pi, 2*np.pi, 400)

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

axs[0, 0].plot(x, np.sin(x))
axs[0, 0].set_title('Sine')

axs[0, 1].plot(x, np.cos(x))
axs[0, 1].set_title('Cosine')

axs[1, 0].plot(x, np.tan(x))
axs[1, 0].set_title('Tangent')
axs[1, 0].set_ylim(-10, 10)  # limit y-axis for tangent

axs[1, 1].plot(x, np.exp(x))
axs[1, 1].set_title('Exponential')

plt.tight_layout()
plt.show()
```
