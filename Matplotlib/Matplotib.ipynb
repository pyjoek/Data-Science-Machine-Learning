```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

presidents_df = pd.read_csv('https://sololearn.com/uploads/files/president_heights_party.csv', index_col='name')
party_cnt = presidents_df['party'].value_counts()

plt.style.use('ggplot')
party_cnt.plot(kind ='bar')
plt.savefig("plot.png")
plt.show()
```

```python
import numpy as np
import pandas as pd

lst = [float(x) if x != 'nan' else np.nan for x in input().split()]
dataFrame = pd.Series(lst)

med = dataFrame.median()

newList = []
for i in dataFrame:
    if np.isnan(i):
        i = med
        newList.append(i)
    else:
        newList.append(i)

newList = pd.Series(newList)
print(newList)
```

```python
import numpy as np
import pandas as pd

lst = [float(x) if x != 'nan' else np.nan for x in input().split()]

series = pd.Series(lst)

mean_value = series.mean()

imputed_series = series.fillna(mean_value)

imputed_series = imputed_series.round(1)

print(imputed_series)
```
