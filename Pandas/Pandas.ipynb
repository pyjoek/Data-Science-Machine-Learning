```python
import numpy as np

r = int(input())
num = [float(x) for x in input().split()]

# Convert to numpy array
arr = np.array(num)

# Reshape the array into r rows
reshaped_arr = arr.reshape(r, -1)

# Round to 2 decimal places
rounded_arr = np.round(reshaped_arr, 2)

print(rounded_arr)
```
